```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
file_path = Path("online-shoppers-dirty.csv")
df = pd.read_csv(file_path, sep=";")
df.drop(columns=["Unnamed: 0"], inplace=True)
```

```python
# fix column names
df.columns = [col.strip().lower() for col in df.columns]
df.columns = df.columns.str.strip()
df.columns = df.columns.str.replace(" ","_")
df.columns = df.columns.str.replace("-","_")
df.columns = df.columns.str.replace("productrelated","product_related")
df.columns = df.columns.str.replace("bouncerates","bounce_rates")
df.columns = df.columns.str.replace("exitrates","exit_rates")
df.columns = df.columns.str.replace("pagevalues","page_values")
df.columns = df.columns.str.replace("specialday","special_day")
df.columns = df.columns.str.replace("operatingsystems","operating_systems")
df.columns = df.columns.str.replace("traffictype","traffic_type")
df.columns = df.columns.str.replace("visitortype","visitor_type")
```
